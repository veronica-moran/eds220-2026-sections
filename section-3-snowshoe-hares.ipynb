# Data

We will use data about Snowshoe hares (Lepus americanus) in the Bonanza Creek Experimental Forest [1].

This dataset is stored in the [Environmental Data Initiative (EDI) data repository](https://portal.edirepository.org/nis/mapbrowse?packageid=knb-lter-bnz.55.22). This is a huge data repository committed to making data Findable, Accessible, Interoperable, and Reusable (FAIR). It is the main repository for all the data associated with the Long Term Ecological Research Network (LTER).

Who assembled this data, and why?
A capture-recapture study was conducted by Kiellant et al., to quantify hare population density in Alaska.

What does a single row in the data represent?
A row contains observation data of a single hare.

When and where were the observations collected?
Observations were collected in Tanana Valley, Alaska from 1999 to 2002.

Does the dataset contain sensitive data?
No.

What study protocols were used to collect the data?
Capture-recapture protocal.

What are the units for the hares’ weight?
Grams

Does the metadata say how missing data is encoded?
NaN

Is there a publication associated with this dataset?
Flora, B.K. 2002. Comparison of snowshoe hare populations in Interior. M.S. Thesis. University of Alaska Fairbanks. Fairbanks, AK, USA.

Citation
Kielland, K., F.S. Chapin, R.W. Ruess, and Bonanza Creek LTER. 2017. Snowshoe hare physical data in Bonanza Creek Experimental Forest: 1999-Present ver 22. Environmental Data Initiative. https://doi.org/10.6073/pasta/03dce4856d79b91557d8e6ce2cbcdc14 (Accessed 2026-10-15).
